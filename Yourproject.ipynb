```python
"""
STUDENT RECORD MANAGEMENT SYSTEM
--------------------------------
Features:
1. Store student data in a file (JSON format)
2. Use dictionary to manage student details
3. Operations:
   - Add student
   - Search student
   - Display all students
   - Update student
   - Delete student
4. Proper file handling and error handling
5. Simple CLI menu-driven interface

Data Fields:
- ID (unique)
- Name
- Age
- Grade
- Marks
"""

import json
import os

FILE_NAME = "students.json"

# ---------------- FILE HANDLING ---------------- #

def load_data():
    """Load student data from file"""
    try:
        if not os.path.exists(FILE_NAME):
            return {}
        with open(FILE_NAME, "r") as file:
            return json.load(file)
    except json.JSONDecodeError:
        print("Error: File is corrupted. Starting with empty data.")
        return {}
    except Exception as e:
        print(f"Unexpected error: {e}")
        return {}


def save_data(data):
    """Save student data to file"""
    try:
        with open(FILE_NAME, "w") as file:
            json.dump(data, file, indent=4)
    except Exception as e:
        print(f"Error saving data: {e}")


# ---------------- CORE FUNCTIONS ---------------- #

def add_student(data):
    """Add a new student"""
    try:
        student_id = input("Enter Student ID: ")
        if student_id in data:
            print("Student already exists!")
            return

        name = input("Enter Name: ")
        age = int(input("Enter Age: "))
        grade = input("Enter Grade: ")
        marks = float(input("Enter Marks: "))

        data[student_id] = {
            "name": name,
            "age": age,
            "grade": grade,
            "marks": marks
        }

        save_data(data)
        print("Student added successfully!\n")

    except ValueError:
        print("Invalid input! Age must be integer and marks must be numeric.")


def search_student(data):
    """Search for a student by ID"""
    student_id = input("Enter Student ID to search: ")
    student = data.get(student_id)

    if student:
        print("\nStudent Found:")
        print(f"ID: {student_id}")
        print(f"Name: {student['name']}")
        print(f"Age: {student['age']}")
        print(f"Grade: {student['grade']}")
        print(f"Marks: {student['marks']}\n")
    else:
        print("Student not found!\n")


def display_students(data):
    """Display all students"""
    if not data:
        print("No student records available.\n")
        return

    print("\n--- All Students ---")
    for sid, student in data.items():
        print(f"ID: {sid} | Name: {student['name']} | Age: {student['age']} | Grade: {student['grade']} | Marks: {student['marks']}")
    print()


def update_student(data):
    """Update existing student details"""
    student_id = input("Enter Student ID to update: ")

    if student_id not in data:
        print("Student not found!\n")
        return

    try:
        print("Leave blank to keep existing value")

        name = input(f"New Name ({data[student_id]['name']}): ")
        age = input(f"New Age ({data[student_id]['age']}): ")
        grade = input(f"New Grade ({data[student_id]['grade']}): ")
        marks = input(f"New Marks ({data[student_id]['marks']}): ")

        if name:
            data[student_id]['name'] = name
        if age:
            data[student_id]['age'] = int(age)
        if grade:
            data[student_id]['grade'] = grade
        if marks:
            data[student_id]['marks'] = float(marks)

        save_data(data)
        print("Student updated successfully!\n")

    except ValueError:
        print("Invalid input!\n")


def delete_student(data):
    """Delete a student"""
    student_id = input("Enter Student ID to delete: ")

    if student_id in data:
        del data[student_id]
        save_data(data)
        print("Student deleted successfully!\n")
    else:
        print("Student not found!\n")


# ---------------- MENU ---------------- #

def menu():
    data = load_data()

    while True:
        print("""
======= STUDENT RECORD SYSTEM =======
1. Add Student
2. Search Student
3. Display All Students
4. Update Student
5. Delete Student
6. Exit
====================================
""")

        choice = input("Enter your choice: ")

        if choice == "1":
            add_student(data)
        elif choice == "2":
            search_student(data)
        elif choice == "3":
            display_students(data)
        elif choice == "4":
            update_student(data)
        elif choice == "5":
            delete_student(data)
        elif choice == "6":
            print("Exiting program...")
            break
        else:
            print("Invalid choice! Please try again.\n")


# ---------------- ENTRY POINT ---------------- #

if __name__ == "__main__":
    menu()
```
